```python
def inspect_request(request_id, dfs):
    # A. Get the request
    requests = dfs["requests"]
    matches = requests[requests["request_id"] == request_id]

    if matches.empty:
        print("Request not found:", request_id)
        return

    request = matches.iloc[0]
    user_id = request["user_id"]

    print("\n=== REQUEST ===")
    print(matches.T.to_string())

    # B. Get the user's financial profile
    profiles = dfs["financial_profiles"]
    profile = profiles[profiles["user_id"] == user_id]

    print("\n=== FINANCIAL PROFILE ===")
    print(profile.T.to_string())

    # C. Get the user's financial events
    events = dfs["financial_events"]
    user_events = events[events["user_id"] == user_id]

    print("\n=== FINANCIAL EVENTS ===")
    print(user_events.to_string(index=False))

    # D. Get payment options for this request
    options = dfs["request_payment_options"]
    request_options = options[
        options["request_id"] == request_id
    ]

    print("\n=== PAYMENT OPTIONS ===")
    print(request_options.to_string(index=False))

    # E. Find relevant messages
    messages = dfs["messages"]
    event_ids = user_events["event_id"]

    related_messages = messages[
        (messages["user_id"] == user_id)
        | (messages["request_id"] == request_id)
        | (messages["related_event_id"].isin(event_ids))
    ]

    print("\n=== MESSAGES ===")
    print(related_messages.to_string(index=False))

    # F. Find relevant image references
    images = dfs["images"]

    related_images = images[
        (images["user_id"] == user_id)
        | (images["request_id"] == request_id)
        | (images["related_event_id"].isin(event_ids))
    ]

    print("\n=== IMAGE REFERENCES ===")
    print(related_images.to_string(index=False))

    # Return the records for further investigation
    return {
        "request": request,
        "profile": profile,
        "events": user_events,
        "payment_options": request_options,
        "messages": related_messages,
        "images": related_images,
    }
```

```python
request_id = dfs["requests"].iloc[0]["request_id"]
context = inspect_request(request_id, dfs)
```

```python
import re
def is_date_format_valid(value):
    if not isinstance(value, str):
        return False

    pattern = r'^\d{2}/\d{2}/\d{4}$'
    return bool(re.fullmatch(pattern, value))
```

# AGENTS

```python
from langgraph.prebuilt import tool_node
from langchain.tools import tool
```

```python
@tool
def get_payment_options(request_id: str):
    """Get all available payment options for a given request ID."""
    df = dfs['request_payment_options']
    result = df[df['request_id'] == request_id]

    return result.to_dict(orient="records")

@tool
def get_user_financial_profile(user_id: str):
    """Get financial profile for a given user ID."""
    df = dfs['financial_profiles']
    result = df[df['user_id'] == user_id]

    return result.to_dict(orient="records")

@tool
def get_past_user_financial_events(user_id: str):
    """Get all past financial events for a given user ID."""
    df = dfs['financial_events']
    result = df[df['user_id'] == user_id]

    return result.to_dict(orient="records")
```

```python
from typing import TypedDict, Any

from langchain_groq import ChatGroq
from langchain_core.messages import SystemMessage, HumanMessage
from langgraph.graph import StateGraph, START, END

import json
import os
from dotenv import load_dotenv
load_dotenv()


# 1. Initialize Groq LLM
llm = ChatGroq(
    model = "openai/gpt-oss-120b",
    temperature = 0,
    api_key = os.environ["GROQ_API_KEY"]
)


# 2. Define graph state
class FinancialAgentState(TypedDict):
    request_id: str
    user_id: str
    payment_options: list[dict[str, Any]]
    financial_profile: list[dict[str, Any]]
    financial_events: list[dict[str, Any]]
    recommendation: str


# 3. Node: retrieve payment options
def payment_options_node(state: FinancialAgentState):
    result = get_payment_options.invoke({
        "request_id": state["request_id"]
    })

    return {"payment_options": result}


# 4. Node: retrieve user's financial profile
def financial_profile_node(state: FinancialAgentState):
    result = get_user_financial_profile.invoke({
        "user_id": state["user_id"]
    })

    return {"financial_profile": result}


# 5. Node: retrieve past financial events
def financial_events_node(state: FinancialAgentState):
    result = get_past_user_financial_events.invoke({
        "user_id": state["user_id"]
    })

    return {"financial_events": result}


# 6. Node: analyze the collected information
def recommendation_node(state: FinancialAgentState):
    payment_options = state["payment_options"]
    financial_profile = state["financial_profile"]
    financial_events = state["financial_events"]

    # Keep all payment options and the user's profile.
    # Limit event details temporarily to reduce prompt size.
    compact_context = {
        "request_id": state["request_id"],
        "user_id": state["user_id"],
        "payment_options": payment_options,
        "financial_profile": financial_profile,
        "financial_events": financial_events[-10:],
        "events_total": len(financial_events),
        "events_included": min(len(financial_events), 10),
    }

    messages = [
        SystemMessage(content="""
You are a financial decision-support agent.

Assess whether the requested expense is affordable.
Use only the supplied evidence.
Do not invent missing data or assume omitted events are irrelevant.
If the supplied history is incomplete, return INSUFFICIENT DATA
when that limitation prevents a reliable decision.

Return BUY, WAIT, or INSUFFICIENT DATA.
Explain the reasoning and payment-method trade-offs.
"""),
        HumanMessage(
            content=json.dumps(compact_context, default=str)
        )
    ]

    response = llm.invoke(messages)

    return {"recommendation": response.content}



# 7. Build graph
graph = StateGraph(FinancialAgentState)

graph.add_node("payment_options", payment_options_node)
graph.add_node("financial_profile", financial_profile_node)
graph.add_node("financial_events", financial_events_node)
graph.add_node("recommendation", recommendation_node)

# Enforce the exact execution order
graph.add_edge(START, "payment_options")
graph.add_edge("payment_options", "financial_profile")
graph.add_edge("financial_profile", "financial_events")
graph.add_edge("financial_events", "recommendation")
graph.add_edge("recommendation", END)

financial_agent = graph.compile()
```
